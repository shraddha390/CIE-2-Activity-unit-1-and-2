```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

from statsmodels.stats.outliers_influence import variance_inflation_factor

print("All libraries imported successfully!")
```

```python
diabetes = load_diabetes(as_frame=True)

df = diabetes.frame

print("Dataset loaded successfully!")
print(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Description:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
X = df[['bmi', 'bp']]
y = df['target']

print("Input Features:")
print(X.head())

print("\nTarget Variable:")
print(y.head())

print("\nInput Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['bmi'], y)
plt.xlabel("BMI")
plt.ylabel("Disease Progression")
plt.title("BMI vs Target")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['bp'], y)
plt.xlabel("BP")
plt.ylabel("Disease Progression")
plt.title("BP vs Target")
plt.show()
```

```python
corr = df[['bmi', 'bp', 'target']].corr()

print(corr)

plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```

```python
sample = df[['bmi', 'bp', 'target']].head(6)

print(sample)

X_small = sample[['bmi', 'bp']].values
y_small = sample['target'].values

# Design matrix
X_design = np.column_stack((
    np.ones(len(X_small)),
    X_small
))

print("\nDesign Matrix X:")
print(X_design)

print("\nTarget Vector Y:")
print(y_small)

# Calculate X transpose X
XTX = X_design.T @ X_design

# Calculate X transpose Y
XTy = X_design.T @ y_small

print("\nX Transpose X:")
print(XTX)

print("\nX Transpose Y:")
print(XTy)

# Solve normal equations
coefficients = np.linalg.solve(XTX, XTy)

print("\nRegression Coefficients:")
print(coefficients)

print("\nIntercept:", coefficients[0])
print("BMI Coefficient:", coefficients[1])
print("BP Coefficient:", coefficients[2])

# Prediction and residual
prediction = X_design @ coefficients

print("\nPredictions:")
print(prediction)

print("\nResiduals:")
print(y_small - prediction)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
X_train_single = X_train[['bmi']]
X_test_single = X_test[['bmi']]

single_model = LinearRegression()

single_model.fit(X_train_single, y_train)

single_pred = single_model.predict(X_test_single)

print("Single Model Intercept:",
      single_model.intercept_)

print("Single Model Coefficient:",
      single_model.coef_)

print("Single Model R2 Score:",
      r2_score(y_test, single_pred))
```

```python
model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Multiple Regression Intercept:",
      model.intercept_)

print("Coefficients:",
      model.coef_)

print("BMI Coefficient:", model.coef_[0])
print("BP Coefficient:", model.coef_[1])
```

```python
manual_model = LinearRegression(fit_intercept=False)

manual_model.fit(X_design, y_small)

print("Manual Calculation Coefficients:")
print(coefficients)

print("\nLibrary Calculation Coefficients:")
print(manual_model.coef_)

print("\nDifference:")
print(coefficients - manual_model.coef_)
```

```python
comparison = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

print(comparison.head(10))
```

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

n = len(y_test)
p = X_test.shape[1]

adjusted_r2 = 1 - (
    (1-r2)*(n-1)/(n-p-1)
)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
print("Adjusted R2 Score:", adjusted_r2)
```

```python
single_mae = mean_absolute_error(y_test, single_pred)
single_mse = mean_squared_error(y_test, single_pred)
single_rmse = np.sqrt(single_mse)
single_r2 = r2_score(y_test, single_pred)

multiple_mae = mean_absolute_error(y_test, y_pred)
multiple_mse = mean_squared_error(y_test, y_pred)
multiple_rmse = np.sqrt(multiple_mse)
multiple_r2 = r2_score(y_test, y_pred)

results = pd.DataFrame({
    'Model': ['Single Linear Regression',
              'Multiple Linear Regression'],
    'MAE': [single_mae, multiple_mae],
    'MSE': [single_mse, multiple_mse],
    'RMSE': [single_rmse, multiple_rmse],
    'R2 Score': [single_r2, multiple_r2]
})

print(results)
```
